```python
import json

with open("indicqa.hi.json", encoding="utf-8") as f:
    obj = json.load(f)

print(type(obj))
```

```python
from pprint import pprint

pprint(obj["data"][0])
```

```python
articles = obj["data"]

passages = [
    p
    for article in articles
    for p in article["paragraphs"]
]

qa_pairs = [
    qa
    for p in passages
    for qa in p["qas"]
]

print("Articles:", len(articles))
print("Passages:", len(passages))
print("Unique passages:", len({p["context"].strip() for p in passages}))
print("QA pairs:", len(qa_pairs))

print("Empty passages:",
      sum(not p.get("context", "").strip() for p in passages))

print("Empty questions:",
      sum(not qa.get("question", "").strip() for qa in qa_pairs))

print("Missing usable answers:",
      sum(
          not any(a.get("text", "").strip() for a in qa.get("answers", []))
          for qa in qa_pairs
      ))
```

```python
shown = 0

for article in obj["data"]:
    for paragraph in article["paragraphs"]:
        for qa in paragraph["qas"]:
            has_answer = any(
                a.get("text", "").strip()
                for a in qa.get("answers", [])
            )

            if not has_answer:
                print("PASSAGE:", paragraph["context"])
                pprint(qa)  # Shows all fields, including any unanswerable flag
                print("-" * 70)

                shown += 1
                if shown == 3:
                    break
        if shown == 3:
            break
    if shown == 3:
        break
```

# creating a csv of 30-50 examples to gauge the odia dataset viability

```python
import csv
import random
from pathlib import Path

# Collect examples that contain answer text.
records = []

for article_idx, article in enumerate(obj["data"]):
    for passage_idx, paragraph in enumerate(article["paragraphs"]):
        for qa_idx, qa in enumerate(paragraph["qas"]):
            answers = [
                a["text"].strip()
                for a in qa.get("answers", [])
                if (a.get("text") or "").strip()
            ]

            if answers:
                records.append({
                    "review_id": f"or_{article_idx}_{passage_idx}_{qa_idx}",
                    "original_id": qa.get("id", ""),
                    "article_id": article_idx,
                    "passage": paragraph["context"],
                    "question": qa["question"],
                    "reference_answers": " || ".join(answers),
                    "verdict": "",
                    "notes": ""
                })

# Fixed seed makes the sample reproducible.
sample = random.Random(42).sample(records, min(40, len(records)))

output = Path("odia_answer_review.csv")

# Avoid overwriting an existing review.
if output.exists():
    raise FileExistsError(
        f"{output} already exists. Rename it before creating another review."
    )

with output.open("w", encoding="utf-8-sig", newline="") as f:
    writer = csv.DictWriter(f, fieldnames=list(sample[0]))
    writer.writeheader()
    writer.writerows(sample)

print(f"Answered examples: {len(records)}")
print(f"Examples to review: {len(sample)}")
print(f"Saved to: {output.resolve()}")
```

```python
import json
import re
import sys
from pathlib import Path
from collections import Counter

INCORRECT = {
    "or_46_0_0", "or_240_0_6", "or_94_0_0",
    "or_140_0_1", "or_106_0_2",
}
UNCERTAIN = {
    "or_57_0_3", "or_244_0_5", "or_229_0_3",
    "or_65_0_0", "or_147_0_4",
}

def normalise(text):
    return re.sub(r"\s+", " ", text).strip()

def save_jsonl(path, rows):
    with path.open("w", encoding="utf-8") as file:
        for row in rows:
            file.write(json.dumps(row, ensure_ascii=False) + "\n")

input_path = Path("/Users/pratyushkumarsethy/Desktop/sem-7/HADL/COURSE-PROJECT/data_and_scripts/indicqa.or.json")
output_dir = Path("/Users/pratyushkumarsethy/Desktop/sem-7/HADL/COURSE-PROJECT/data_and_scripts/odia_real_cleaned")
output_dir.mkdir(parents=True, exist_ok=True)

data = json.loads(input_path.read_text(encoding="utf-8-sig"))
candidates = []
needs_review = []
seen = set()
reasons_count = Counter()

for article_i, article in enumerate(data["data"]):
    for passage_i, paragraph in enumerate(article["paragraphs"]):
        passage = paragraph["context"].strip()

        for question_i, qa in enumerate(paragraph["qas"]):
            review_id = f"or_{article_i}_{passage_i}_{question_i}"
            question = qa["question"].strip()
            answers = [
                answer.get("text", "").strip()
                for answer in qa.get("answers", [])
            ]
            answers = [answer for answer in answers if answer]

            row = {
                "review_id": review_id,
                "original_id": qa.get("id"),
                "article_id": article_i,
                "passage_id": f"or_{article_i}_{passage_i}",
                "category": qa.get("category"),
                "passage": passage,
                "question": question,
                "answers": answers,
            }

            reasons = []

            if not passage or not question:
                reasons.append("empty_passage_or_question")
            if not answers:
                reasons.append("blank_answer")
            elif not any(answer in passage for answer in answers):
                reasons.append("answer_not_found_in_passage")
            if review_id in INCORRECT:
                reasons.append("incorrect_in_review")
            if review_id in UNCERTAIN:
                reasons.append("uncertain_in_review")

            duplicate_key = (normalise(passage), normalise(question))
            if duplicate_key in seen:
                reasons.append("duplicate_passage_question")
            else:
                seen.add(duplicate_key)

            if reasons:
                row["review_reasons"] = reasons
                needs_review.append(row)
                reasons_count.update(reasons)
            else:
                candidates.append(row)

save_jsonl(output_dir / "candidate_qa.jsonl", candidates)
save_jsonl(output_dir / "needs_review.jsonl", needs_review)

print("Candidate QA pairs:", len(candidates))
print("Set aside for review:", len(needs_review))
print("Reasons:", dict(reasons_count))
```

# creating train/test/validation set 70/15/15 

answers pertaining to one passage in one particular set

```python
import json
import random
from collections import defaultdict
from pathlib import Path

folder = Path("/Users/pratyushkumarsethy/Desktop/sem-7/HADL/COURSE-PROJECT/data_and_scripts/odia_real_cleaned")

rows = [
    json.loads(line)
    for line in (folder / "candidate_qa.jsonl").read_text(encoding="utf-8").splitlines()
]

articles = defaultdict(list)
for row in rows:
    articles[row["article_id"]].append(row)

article_ids = list(articles)
random.Random(42).shuffle(article_ids)

n_train = round(0.70 * len(article_ids))
n_val = round(0.15 * len(article_ids))

split_ids = {
    "train": article_ids[:n_train],
    "validation": article_ids[n_train:n_train + n_val],
    "test": article_ids[n_train + n_val:],
}

for split_name, ids in split_ids.items():
    output = folder / f"{split_name}.jsonl"
    with output.open("w", encoding="utf-8") as file:
        for article_id in ids:
            for row in articles[article_id]:
                file.write(json.dumps(row, ensure_ascii=False) + "\n")
    print(split_name, "articles:", len(ids), "questions:", sum(len(articles[i]) for i in ids))
```
